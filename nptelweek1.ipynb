```r
# LECTURE 3: Packages and Libraries in R[cite: 3]
# ==============================================================================

# Installing new packages from CRAN[cite: 3]
install.packages("boot")
install.packages("cluster")

# Loading an installed library for use[cite: 3]
library(spatial)

# Getting the description file details of a package[cite: 3]
packageDescription("spatial")

# Discovering the contents and functions of a library[cite: 3]
library(help=spatial)

# Checking which packages are currently installed on your computer[cite: 3]
installed.packages()

# Updating an installed package[cite: 3]
update.packages("cluster")

# --- THE FIX ---
# Loading the package into the active environment FIRST so it can be detached
library(cluster)

# Unloading the active package from the current session[cite: 3]
detach("package:cluster", unload=TRUE)

# Uninstalling/removing a package from your computer entirely[cite: 3]
remove.packages("cluster")
```
