```r
# Checking numbers and characters[cite: 4]
x = 20
is.numeric(x)
is.character(x)

y = "apple"
is.character(y)
is.numeric(y)

```

```r
# Converting between numbers and characters[cite: 4]
x = 20
is.numeric(x)
y = as.character(x)
is.numeric(y)
is.character(y)
y

y = "apple"
is.numeric(y)
is.character(y)
z = as.numeric(y)
is.numeric(z)
is.character(z)
z

```

```r
# LECTURE 9: Calculations with Data Vectors (Vector vs Vector)[cite: 4]

# Addition[cite: 4]
c(2, 3, 5, 7) + c(-2, -3, -5, 8)
c(2, 3, 5, 7) + c(8, 9)
c(2, 3, 5, 7) + c(8, 9, 10)

# Subtraction[cite: 4]
c(2, 3, 5, 7) - c(-2, -3, -5, 8)
c(12, 13, 15, 17) - c(8, 9)
c(12, 13, 15, 17) - c(8, 9, 10)

# Multiplication[cite: 4]
c(2, 3, 5, 7) * c(-2, -3, -5, 8)
c(2, 3, 5, 7) * c(8, 9)
c(2, 3, 5, 7) * c(8, 9, 10)

# Division[cite: 4]
c(24, 20, 8, 16) / c(3, 4, 2, 8)
c(24, 20, 8, 16) / c(4, 2)
c(24, 20, 8, 16) / c(4, 2, 8)
```
