```r
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)

labels.dieface <- c(
  "one", "two", "three",
  "four", "five", "six"
)
```

```r
facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

print(facy)
```

```r
levels(facy)
```

```r
# Vector
x <- 1:10
print(x)

# Vector indexing
print(x[x > 5])
print(x[x %% 2 == 0])

# List
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

# Merge lists
list12 <- c(list1, list2)
print(list12)

# Convert list to vector
print(unlist(list1))

# Factor
gender <- factor(c("Male", "Female", "Female", "Male"))
print(gender)
print(levels(gender))

# Named vector
drinks <- c(water = 1, juice = 2, lemonade = 3)
print(drinks["juice"])

# String formatting
print("Hello R")

# Format number
print(format(0.5, digits = 10, nsmall = 5))

# Large number formatting
print(format(123456789, big.mark = ","))

# Matrix
m <- matrix(1:6, nrow = 3, ncol = 2, byrow = TRUE)
print(m)
```
