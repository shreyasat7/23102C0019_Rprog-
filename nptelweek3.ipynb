```r
# LECTURE 10: Power Operations, Integer & Modulo Divisions[cite: 2]

# Power operators (^ and **) with scalars[cite: 2]
2^3
2**3
2^0.5
2**0.5
2^-0.5
```

```r
# Power operations with vectors[cite: 2]
c(2, 3, 5, 7)^2
c(2, 3, 5, 7)^c(2, 3)
c(1, 2, 3, 4, 5, 6)^c(2, 3, 4)
c(2, 3, 5, 7)^c(2, 3, 4) # Recycled with warning[cite: 2]
```

```r
# Integer Division (%/%)[cite: 2]
2 %/% 2
5 %/% 2
7 %/% 3
c(2, 3, 5, 7) %/% 2
c(2, 3, 5, 7) %/% c(2, 3)
c(2, 3, 5) %/% c(2, 3) # Recycled with warning[cite: 2]
```

```r
# Modulo Division (%%)[cite: 2]
2 %% 2
3 %% 2
7 %% 3
7 %% 4
c(2, 3, 5, 7) %% 2
c(2, 3, 5, 7) %% c(2, 3)
c(2, 3, 5) %% c(2, 3) # Recycled with warning[cite: 2]
```
