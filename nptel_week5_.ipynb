```r
#LECTURE 20: Nested If-Else and Ifelse()[cite: 1]

x = 5
if ( x == 3 ) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else { x = 2 * x }
x

x = 2
if ( x == 3 ) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else { x = 2 * x }
x

x = 3
if ( x == 3 ) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else { x = 2 * x }
x

x = 1:10
x
ifelse(x < 6, x^2, x + 1)

x = c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
```
